```python
def categorizar_veiculo(tipo):
    tipo_lower = str(tipo).lower().strip()
    if tipo_lower in ['automóvel', 'utilitário', 'camioneta', 'caminhonete', 'motor-casa']:
        return 'Transporte Utilitário e Familiar'
    elif tipo_lower in ['ônibus', 'micro-ônibus', 'bonde / trem', 'trem-bonde']:
        return 'Transporte Coletivo'
    elif tipo_lower in ['motocicletas', 'ciclomotor', 'motoneta', 'motocicleta','triciclo','quadriciclo']:
        return 'Duas e Quatro Rodas Leves'
    elif tipo_lower in ['caminhão', 'caminhão-trator', 'semi-reboque', 'caminhão-tanque', 'reboque', 'chassi-plataforma']:
        return 'Carga e Transporte Pesado'
    elif tipo_lower == 'bicicleta':
        return 'Não Motorizados'
    elif tipo_lower in ['charrete', 'carroça', 'carro-de-mão', 'carroça-charrete', 'outros']:
        return 'Tração Animal e Outros'
    elif tipo_lower in ['trator de rodas', 'trator de esteiras', 'trator misto', 'trator de esteira']:
        return 'Máquinas Agrícolas e de Construção'
    elif tipo_lower in ['não informado', 'não identificado']:
        return 'Erros ou dados faltantes'
    else:
        return 'Outros/Não Classificado'

PRF_FINAL['categorias'] = PRF_FINAL['tipo_veiculo'].apply(categorizar_veiculo)

print("Coluna 'categorias' criada com sucesso para PRF_FINAL.")
print("Novas categorias em PRF_FINAL:")
print(PRF_FINAL['categorias'].unique())
display(PRF_FINAL.head())

# Aplicar a categorização ao PRF_FINAL_15_16
PRF_FINAL_15_16['categorias'] = PRF_FINAL_15_16['tipo_veiculo'].apply(categorizar_veiculo)

print("\nColuna 'categorias' criada com sucesso para PRF_FINAL_15_16.")
print("Novas categorias em PRF_FINAL_15_16:")
print(PRF_FINAL_15_16['categorias'].unique())
display(PRF_FINAL_15_16.head())
```

```python
MAX_AGE = 100

# Função para tratar idades irrealistas
def tratar_idade_irrealista(df_name, df):
    print(f"Verificando e corrigindo valores de 'idade' no DataFrame: {df_name}")
    if 'idade' in df.columns:
        # Substituir idades maiores que MAX_AGE ou menores que 0 por NaN
        # O fillna subsequente (já implementado em outras células) converterá NaN para 0
        outliers_high = df[df['idade'] > MAX_AGE]['idade'].count()
        outliers_low = df[df['idade'] < 0]['idade'].count()

        if outliers_high > 0:
            df.loc[df['idade'] > MAX_AGE, 'idade'] = pd.NA
            print(f"  - {outliers_high} valores de 'idade' acima de {MAX_AGE} foram substituídos por NA em {df_name}.")
        else:
            print(f"  - Nenhuma idade acima de {MAX_AGE} encontrada em {df_name}.")

        if outliers_low > 0:
            df.loc[df['idade'] < 0, 'idade'] = pd.NA
            print(f"  - {outliers_low} valores de 'idade' abaixo de 0 (incluindo -1) foram substituídos por NA em {df_name}.")
        else:
            print(f"  - Nenhuma idade abaixo de 0 encontrada em {df_name}.")

    else:
        print(f"  - Coluna 'idade' não encontrada em {df_name}.")

# Aplicar a função a ambos os DataFrames
tratar_idade_irrealista('PRF_FINAL', PRF_FINAL)
tratar_idade_irrealista('PRF_FINAL_15_16', PRF_FINAL_15_16)

print("Processo de correção de idades irrealistas concluído. Lembre-se de re-executar as células de `fillna` e `convert_float_to_int_if_applicable` caso não tenha feito após esta alteração para garantir que os NAs sejam convertidos para 0 e o tipo Int64Dtype seja aplicado.")
```

```python
import pandas as pd

# Lista de colunas que geralmente são números inteiros, mas podem ser floats com '.0'
# Excluindo 'km' conforme solicitado.
integer_like_columns_to_convert = [
    'id', 'pesid', 'id_veiculo', 'ano_fabricacao_veiculo', 'idade',
    'mortos', 'feridos_leves', 'feridos_graves', 'ilesos', 'ignorados',
    'feridos', 'veiculos', 'br', 'ano'
]

def convert_float_to_int_if_applicable(df_name, df):
    print(f"Processando o DataFrame: {df_name}")
    for col in integer_like_columns_to_convert:
        if col in df.columns:
            # Verifica se a coluna é numérica (float) e pode ser convertida para inteiro
            # Usamos pd.Int64Dtype() para permitir valores NaN (missing values) em colunas inteiras
            if pd.api.types.is_float_dtype(df[col]):
                df[col] = df[col].astype(pd.Int64Dtype())
                print(f"  - Coluna '{col}' em {df_name} convertida para Int64Dtype.")
    print(f"Colunas numéricas com '.0' removido em {df_name} (convertidas para tipo inteiro, se aplicável).")

# Aplicar a função para os DataFrames
convert_float_to_int_if_applicable('PRF_FINAL', PRF_FINAL)
convert_float_to_int_if_applicable('PRF_FINAL_15_16', PRF_FINAL_15_16)

print("\nVerifique o .head() ou .info() dos DataFrames para confirmar as alterações e que 'km' manteve seu tipo original.")
```

```python
# Processando o DataFrame PRF_FINAL_15_16 para a coluna 'data_inversa'
if 'data_inversa' in PRF_FINAL_15_16.columns:
    # Tentar converter usando um formato padrão (e.g., YYYY-MM-DD) ou inferir se possível
    PRF_FINAL_15_16['data'] = pd.to_datetime(PRF_FINAL_15_16['data_inversa'], errors='coerce', format='%Y-%m-%d')

    # Identificar valores que falharam na primeira tentativa
    mask_na_15_16 = PRF_FINAL_15_16['data'].isna()

    # Tentar um formato alternativo para os valores que falharam (e.g., DD/MM/YYYY)
    PRF_FINAL_15_16.loc[mask_na_15_16, 'data'] = pd.to_datetime(
        PRF_FINAL_15_16.loc[mask_na_15_16, 'data_inversa'],
        errors='coerce',
        format='%d/%m/%Y'
    )

    # Após a conversão, remover a coluna 'data_inversa' original
    PRF_FINAL_15_16 = PRF_FINAL_15_16.drop(columns=['data_inversa'])
    print("Coluna 'data_inversa' convertida para o formato de data padrão e renomeada para 'data' em PRF_FINAL_15_16.")
else:
    print("Coluna 'data_inversa' não encontrada em PRF_FINAL_15_16. É possível que já tenha sido processada ou renomeada.")
    if 'data' in PRF_FINAL_15_16.columns:
        print("A coluna 'data' já existe em PRF_FINAL_15_16. Nenhuma ação adicional é necessária para conversão de data.")
    else:
        print("Erro: Nem 'data_inversa' nem 'data' foram encontradas em PRF_FINAL_15_16. Por favor, verifique os nomes das colunas.")
```

```python
def mapear_gravidade_geral(estado_fisico):
    estado_fisico_lower = str(estado_fisico).lower().strip()
    if estado_fisico_lower == 'óbito' or estado_fisico_lower == 'morto':
        return 'Morte'
    elif estado_fisico_lower == 'lesões graves' or estado_fisico_lower == 'ferido grave':
        return 'Ferido Grave'
    elif estado_fisico_lower == 'lesões leves' or estado_fisico_lower == 'ferido leve':
        return 'Ferido Leve'
    elif estado_fisico_lower == 'ileso':
        return 'Ileso'
    elif estado_fisico_lower in ['não informado', '(null)', 'ignorado']:
        return 'Não Informado'
    else:
        return 'Outros'

# Exibir os valores únicos e suas contagens para 'estado_fisico' antes de aplicar o mapeamento
print("Valores únicos e suas contagens para 'estado_fisico' em PRF_FINAL_15_16:")
print(PRF_FINAL_15_16['estado_fisico'].value_counts(dropna=False))

# Criar a coluna 'gravidade_geral' em PRF_FINAL_15_16
PRF_FINAL_15_16['gravidade_geral'] = PRF_FINAL_15_16['estado_fisico'].apply(mapear_gravidade_geral)
print("\nColuna 'gravidade_geral' criada com sucesso em PRF_FINAL_15_16, com lógica de categorização aprimorada.")

# Exibir as primeiras linhas e contagem de valores para PRF_FINAL_15_16
print("\nPrimeiras linhas de PRF_FINAL_15_16 com a nova coluna 'gravidade_geral':")
display(PRF_FINAL_15_16[['estado_fisico', 'gravidade_geral']].head())
print("Contagem de valores para 'gravidade_geral' em PRF_FINAL_15_16:")
print(PRF_FINAL_15_16['gravidade_geral'].value_counts(dropna=False))

# Adicionar uma verificação explícita para a categoria 'Outros'
if 'Outros' in PRF_FINAL_15_16['gravidade_geral'].unique():
    print(f"ATENÇÃO: A categoria 'Outros' ainda existe na coluna 'gravidade_geral' com {PRF_FINAL_15_16['gravidade_geral'].value_counts()['Outros']} ocorrências. Isso pode indicar valores de `estado_fisico` não mapeados.")
else:
    print("\nNenhuma ocorrência da categoria 'Outros' foi encontrada na coluna 'gravidade_geral'. Todos os valores de 'estado_fisico' foram mapeados para as categorias definidas.")
```

```python
# Padronizar a coluna 'estado_fisico' para remover espaços e capitalização inconsistente
PRF_FINAL_15_16['estado_fisico'] = PRF_FINAL_15_16['estado_fisico'].astype(str).str.strip().str.lower()

print("Coluna 'estado_fisico' padronizada (removendo espaços e convertendo para minúsculas).")

# Exibir os valores únicos e suas contagens após a padronização
print("\nValores únicos e suas contagens para 'estado_fisico' em PRF_FINAL_15_16 (APÓS padronização):")
print(PRF_FINAL_15_16['estado_fisico'].value_counts(dropna=False))

# Reaplicar a função de mapeamento, caso a padronização tenha alterado algum valor que afetasse o mapeamento original
# Isso garante que a coluna 'gravidade_geral' reflita a versão mais limpa de 'estado_fisico'
PRF_FINAL_15_16['gravidade_geral'] = PRF_FINAL_15_16['estado_fisico'].apply(mapear_gravidade_geral)
print("\nColuna 'gravidade_geral' atualizada após a padronização de 'estado_fisico'.")
print("Contagem de valores para 'gravidade_geral' em PRF_FINAL_15_16 (APÓS atualização):")
print(PRF_FINAL_15_16['gravidade_geral'].value_counts(dropna=False))

# Demonstração de como ver apenas os valores de uma categoria específica
categoria_selecionada = 'Morte' # Altere esta variável para ver outras categorias, como 'Ileso', 'Ferido Leve', 'Ferido Grave', 'Não Informado'

print(f"\nExibindo apenas a contagem para a categoria '{categoria_selecionada}' na coluna 'gravidade_geral':")
contagem_categoria_selecionada = PRF_FINAL_15_16[PRF_FINAL_15_16['gravidade_geral'] == categoria_selecionada]['gravidade_geral'].value_counts()
print(contagem_categoria_selecionada)

print(f"\nPara ver outras categorias, basta alterar a variável `categoria_selecionada` (por exemplo, para 'Ileso', 'Ferido Leve', 'Ferido Grave', 'Não Informado').")
```

```python
# Identificar colunas com alta porcentagem de valores 'Não Informado' ou nulos

threshold = 0.8 # Definir um limite, por exemplo, 80% de valores vazios/Não Informado

print("Verificando colunas com alta porcentagem de valores 'Não Informado' ou nulos em PRF_FINAL_15_16:")

for col in PRF_FINAL_15_16.columns:
    # Calcular a porcentagem de NaNs
    nan_percentage = PRF_FINAL_15_16[col].isnull().sum() / len(PRF_FINAL_15_16) if not PRF_FINAL_15_16[col].empty else 0

    # Calcular a porcentagem de 'Não Informado' (case-insensitive e trimmed)
    nao_informado_count = PRF_FINAL_15_16[col].astype(str).str.strip().str.lower().eq('não informado').sum()
    nao_informado_percentage = nao_informado_count / len(PRF_FINAL_15_16) if not PRF_FINAL_15_16[col].empty else 0

    # Se a soma das porcentagens ultrapassar o threshold, imprimir
    if (nan_percentage + nao_informado_percentage) > threshold:
        print(f"  - Coluna '{col}': {nan_percentage:.2%} NaN, {nao_informado_percentage:.2%} 'Não Informado'. Total de valores ausentes/não informados: {(nan_percentage + nao_informado_percentage):.2%}")

print("\nSe alguma coluna for listada acima, ela pode ser considerada 'vazia' e podemos discutir a remoção dela se não for relevante para sua análise no Power BI.")
```
