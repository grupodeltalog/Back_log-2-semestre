```python
import pandas as pd

from google.colab import drive

drive.mount ('/content/drive')

origem = '/content/drive/My Drive/datasus_onsv/'
```

```python
arquivo_1 ='DOEXT15.csv'
arquivo_2 ='DOEXT16.csv'
arquivo_3 ='DOEXT17.csv'
arquivo_4 ='DOEXT18.csv'
arquivo_5 ='DOEXT19.csv'
arquivo_6 ='DOEXT20.csv'
arquivo_7 ='DOEXT21.csv'
arquivo_8 ='DOEXT22.csv'
arquivo_9 ='DOEXT23.csv'
arquivo_10 ='DOEXT24.csv'
arquivo_11 ='DOEXT25.csv'
```

```python
dt_2015 = pd.read_csv(origem + arquivo_1, low_memory=False, sep=',', encoding='latin1')
dt_2016 = pd.read_csv(origem + arquivo_2, low_memory=False, sep=',', encoding='latin1')
dt_2017 = pd.read_csv(origem + arquivo_3, low_memory=False, sep=',', encoding='latin1')
dt_2018 = pd.read_csv(origem + arquivo_4, low_memory=False, sep=',', encoding='latin1')
dt_2019 = pd.read_csv(origem + arquivo_5, low_memory=False, sep=',', encoding='latin1')
dt_2020 = pd.read_csv(origem + arquivo_6, low_memory=False, sep=',', encoding='latin1')
dt_2021 = pd.read_csv(origem + arquivo_7, low_memory=False, sep=',', encoding='latin1')
dt_2022 = pd.read_csv(origem + arquivo_8, low_memory=False, sep=',', encoding='latin1')
df_2023 = pd.read_csv(origem + arquivo_9, low_memory=False, sep=',', encoding='latin1')
dt_2024 = pd.read_csv(origem + arquivo_10, low_memory=False, sep=',', encoding='latin1')
dt_2025 = pd.read_csv(origem + arquivo_11, low_memory=False, sep=',', encoding='latin1')
```

```python
dt_final = pd.concat([dt_2015, dt_2016, dt_2017, dt_2018, dt_2019, dt_2020, dt_2021, dt_2022, df_2023, dt_2024, dt_2025])
```
