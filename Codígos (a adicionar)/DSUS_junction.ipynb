```python
import pandas as pd

from google.colab import drive

drive.mount ('/content/drive')

origem = '/content/drive/My Drive/datasus/cnvtocsv/'
```

```python
arquivo_1 = "dt_2015.csv"
arquivo_2 = "dt_2025.csv"
```

```python
dt1 = pd.read_csv(origem + arquivo_1, low_memory=False, sep=';', encoding='UTF-8')
dt2 = pd.read_csv(origem + arquivo_2, low_memory=False, sep=';', encoding='UTF-8')
```

```python
dt_final = pd.concat([dt1, dt2], ignore_index=True)
```

```python
padrao_transito = 'acidente|transito|trânsito|colisao|colisão|motociclista|atropelamento|veiculo|veículo|automovel|automóvel|condutor|pedestre'

dt_final = dt_final[dt_final['CAUSABAS'].str.contains(padrao_transito, case=False, na=False)]
```

```python
dt_final.tail()
```

```python
caminho_salvamento = origem + 'dt_final.csv'

dt_final.to_csv(caminho_salvamento, index=False, encoding='utf-8')

print(f"Arquivo salvo com todas as alterações com sucesso em: {caminho_salvamento}")
```
