```python
import pandas as pd

from google.colab import drive

drive.mount ('/content/drive')

origem = '/content/drive/My Drive/relação/'
```

```python
arquivo_1 = "PRF_FINAL_15_16.csv"
arquivo_2 = "DATASUS.csv"
```

```python
prf = pd.read_csv(origem + arquivo_1, low_memory=False, sep=',', encoding='latin1')
dtsus = pd.read_csv(origem + arquivo_2, low_memory=False, sep=',', encoding='latin1')
```

```python
# Exibindo informações gerais das colunas do DATASUS
print('--- DATASUS Columns & Info ---')
display(dtsus.info())
display(dtsus.head(3))
```

```python
# Exibindo informações gerais das colunas da PRF
print('--- PRF Columns & Info ---')
display(prf.info())
display(prf.head(3))
```

```python
# 1. Padronização de datas
# DATASUS: DTOBITO está como inteiro (ex: 12042015 -> 12/04/2015)
def formatar_data_dtsus(val):
    s = str(val).zfill(8)
    if len(s) == 8:
        return f"{s[4:8]}-{s[2:4]}-{s[0:2]}"
    return None

dtsus['data_padrao'] = dtsus['DTOBITO'].apply(formatar_data_dtsus)
prf['data_padrao'] = pd.to_datetime(prf['data'], errors='coerce').dt.strftime('%Y-%m-%d')

# 2. Padronização de Idade
# DATASUS: '53 Anos' -> 53
dtsus['idade_padrao'] = dtsus['IDADE'].str.extract('(\d+)').astype(float)
prf['idade_padrao'] = prf['idade'].astype(float)

# 3. Padronização de Sexo
dtsus['sexo_padrao'] = dtsus['SEXO'].astype(str).str.strip().str.upper()
prf['sexo_padrao'] = prf['sexo'].astype(str).str.strip().str.upper()

# 4. Realizando o merge (cruzamento) aproximado com base nas chaves padronizadas
dados_cruzados = pd.merge(
    prf,
    dtsus,
    on=['data_padrao', 'sexo_padrao', 'idade_padrao'],
    how='inner'
)

print(f"Quantidade de registros unificados encontrados: {dados_cruzados.shape[0]}")
display(dados_cruzados.head())
```

```python
# Configurar o pandas para não ocultar colunas ao exibir
pd.set_option('display.max_columns', None)

# Exibir a lista completa de colunas unificadas
print("Colunas no DataFrame unificado (dados_cruzados):")
for i, col in enumerate(dados_cruzados.columns, 1):
    print(f"{i}. {col}")

# Exibir as primeiras 3 linhas com todas as colunas visíveis
display(dados_cruzados.head(10))
```
