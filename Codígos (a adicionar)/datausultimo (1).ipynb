```python
import pandas as pd

from google.colab import drive

drive.mount ('/content/drive')

origem = '/content/drive/My Drive/datasus/'
```

```python
arquivo_1 = "dt_final_c.csv"
```

```python
dt_final_c = pd.read_csv(origem + arquivo_1, low_memory=False, sep=',', encoding='latin1')
```

```python
pd.set_option('display.max_columns', None)
dt_final_c.head()
```

```python
# Definindo palavras-chave em minúsculo relacionadas a acidentes de trânsito
termos_transito = ['transp', 'motociclista', 'pedestre', 'colisao', 'veiculo', 'atropelamento', 'transito', 'ciclista', 'automovel', 'condutor', 'ocupante']

# Criando uma máscara para filtrar a coluna CAUSABAS
mascara = dt_final_c['CAUSABAS'].astype(str).str.lower().str.contains('|'.join(termos_transito), na=False)

# Aplicando o filtro
dt_final_c = dt_final_c[mascara]
dt_final_c.head(1)
```

```python
# Remover a coluna TIPOBITO
dt_final_c = dt_final_c.drop(columns=['TIPOBITO'], errors='ignore')

# Substituir NaN por 'Não Informado'
dt_final_c = dt_final_c.fillna('Não Informado')

# Visualizar o resultado
dt_final_c.head(10)
```

```python
# Removendo a coluna original CODMUNRES
dt_final_c = dt_final_c.drop(columns=['CODMUNRES', 'RACACOR', 'NECROPSIA', 'LINHAD', 'LINHAII', 'fonte_arquivo', 'ESTCIV'], errors='ignore')

# Exibindo as primeiras linhas do DataFrame atualizado
display(dt_final_c.head(5))
```
