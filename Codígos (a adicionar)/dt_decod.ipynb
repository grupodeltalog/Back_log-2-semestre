```python
import os
import pandas as pd

def ler_arquivo_cnv(caminho_arquivo):
    """
    Lê um arquivo .cnv do DATASUS e retorna um dicionário mapeando os códigos de CID para suas descrições.
    """
    mapeamento = {}
    try:
        with open(caminho_arquivo, 'r', encoding='latin1') as f:
            linhas = f.readlines()

        # Processa cada linha procurando padrões de conversão
        for linha in linhas:
            linha = linha.strip()
            # Ignora cabeçalhos ou linhas vazias
            if not linha or linha.startswith(';') or linha.upper().startswith('MAPA') or linha.upper().startswith('SEM'):
                continue

            # Linhas válidas do CNV geralmente têm o formato: CODIGO DESCRICAO ou CODIGO;DESCRICAO
            parts = linha.split(' ', 1) if ' ' in linha else linha.split(';', 1)
            if len(parts) == 2:
                codigo = parts[0].strip().replace('"', '')
                descricao = parts[1].strip().replace('"', '')
                mapeamento[codigo] = descricao
    except Exception as e:
        print(f"Erro ao ler o arquivo CNV: {e}")
    return mapeamento

# Caminho do arquivo de CID-10 em 3 dígitos
caminho_cnv = os.path.join(pasta_tabdo, 'CID10_3D.CNV')

if os.path.exists(caminho_cnv):
    dic_cid = ler_arquivo_cnv(caminho_cnv)
    print(f"Sucesso! {len(dic_cid)} termos de CID mapeados do arquivo CNV.")

    # Teste de transcrição no DataFrame para a coluna CAUSABAS
    # Como CAUSABAS muitas vezes tem 4 caracteres (ex: V299), pegamos os 3 primeiros caracteres para bater com o CID10_3D
    dt_final['CAUSABAS_3DIG'] = dt_final['CAUSABAS'].str[:3]
    dt_final['CAUSA_OBITO_DESC'] = dt_final['CAUSABAS_3DIG'].map(dic_cid)

    # Exibe amostra do mapeamento
    display(dt_final[['CAUSABAS', 'CAUSABAS_3DIG', 'CAUSA_OBITO_DESC']].dropna().head(10))
else:
    print(f"O arquivo {caminho_cnv} não foi encontrado.")
```

```python
def ler_arquivo_cnv_correto(caminho_arquivo):
    """
    Lê o arquivo .cnv tratando a estrutura interna para extrair o código CID
    e sua respectiva descrição textual.
    """
    mapeamento = {}
    try:
        with open(caminho_arquivo, 'r', encoding='latin1') as f:
            linhas = f.readlines()

        for linha in linhas:
            linha = linha.strip()
            if not linha or linha.startswith(';') or linha.upper().startswith('MAPA') or linha.upper().startswith('SEM'):
                continue

            # Divide a linha: ex '1 A00   Colera                                       A00,'
            # Queremos extrair 'A00' e 'Colera'
            partes_virgula = linha.split(',')
            parte_principal = partes_virgula[0].strip()

            # Divide por múltiplos espaços
            componentes = [c.strip() for c in parte_principal.split('  ') if c.strip()]

            if len(componentes) >= 3:
                # Ex: ['1', 'A00', 'Colera']
                codigo_cid = componentes[1]
                descricao = componentes[2]
                mapeamento[codigo_cid] = descricao
            elif len(componentes) == 2:
                # Ex caso simplificado: ['A00', 'Colera']
                codigo_cid = componentes[0]
                descricao = componentes[1]
                mapeamento[codigo_cid] = descricao
    except Exception as e:
        print(f"Erro ao processar: {e}")
    return mapeamento

# Gerando o novo mapeamento ajustado
dic_cid_correto = ler_arquivo_cnv_correto(caminho_cnv)
print(f"Novos termos mapeados corretamente: {len(dic_cid_correto)}")

# Vamos ver um exemplo do dicionário processado:
exemplo_chaves = list(dic_cid_correto.keys())[:5]
print("Exemplo de mapeamento correto:", {k: dic_cid_correto[k] for k in exemplo_chaves})

# Aplicando a transcrição oficial no DataFrame
dt_final['CAUSABAS_3DIG'] = dt_final['CAUSABAS'].str[:3]
dt_final['CAUSA_OBITO_DESC'] = dt_final['CAUSABAS_3DIG'].map(dic_cid_correto)

# Exibe o resultado final com as traduções das causas
display(dt_final[['CAUSABAS', 'CAUSABAS_3DIG', 'CAUSA_OBITO_DESC']].dropna().head(10))
```

### Etapa 1: Definição de Funções e Mapeamentos Estáticos
Nesta etapa, definimos a função padrão para interpretar os arquivos dicionários do DATASUS (`.CNV`) e criamos as tabelas de tradução para as colunas que possuem mapeamento fixo estruturado.

```python
import os
import pandas as pd
import numpy as np

def ler_arquivo_cnv_correto(caminho_arquivo):
    """
    Lê arquivos de dicionário .CNV do DATASUS interpretando corretamente
    a estrutura de colunas internas.
    """
    mapeamento = {}
    try:
        with open(caminho_arquivo, 'r', encoding='latin1') as f:
            linhas = f.readlines()
        for linha in linhas:
            linha = linha.strip()
            if not linha or linha.startswith(';') or linha.upper().startswith('MAPA') or linha.upper().startswith('SEM'):
                continue
            partes_virgula = linha.split(',')
            parte_principal = partes_virgula[0].strip()
            componentes = [c.strip() for c in parte_principal.split('  ') if c.strip()]
            if len(componentes) >= 3:
                mapeamento[componentes[1]] = componentes[2]
            elif len(componentes) == 2:
                mapeamento[componentes[0]] = componentes[1]
    except Exception as e:
        print(f"Erro ao processar {caminho_arquivo}: {e}")
    return mapeamento

# --- Mapeamentos Estáticos ---
dic_sexo = {1: 'Masculino', 2: 'Feminino', 0: 'Ignorado', '1': 'Masculino', '2': 'Feminino', '0': 'Ignorado'}

dic_racacor = {
    1: 'Branca', 2: 'Preta', 3: 'Amarela', 4: 'Parda', 5: 'Indígena', 9: 'Ignorado',
    '1': 'Branca', '2': 'Preta', '3': 'Amarela', '4': 'Parda', '5': 'Indígena', '9': 'Ignorado'
}

dic_lococor = {
    1: 'Hospital', 2: 'Outros estabelecimentos de saúde', 3: 'Domicílio',
    4: 'Via pública', 5: 'Outros', 9: 'Ignorado',
    '1': 'Hospital', '2': 'Outros estabelecimentos de saúde', '3': 'Domicílio',
    '4': 'Via pública', '5': 'Outros', '9': 'Ignorado'
}

dic_tipobito = {1: 'Fetal', 2: 'Não Fetal', '1': 'Fetal', '2': 'Não Fetal'}

dic_esc2010 = {
    '0': 'Sem escolaridade', '1': 'Fundamental I (1ª a 4ª série)', '2': 'Fundamental II (5ª a 8ª série)',
    '3': 'Ensino Médio', '4': 'Ensino Superior incompleto/completo', '5': 'Ensino Superior completo', '9': 'Ignorado',
    0: 'Sem escolaridade', 1: 'Fundamental I (1ª a 4ª série)', 2: 'Fundamental II (5ª a 8ª série)',
    3: 'Ensino Médio', 4: 'Ensino Superior incompleto/completo', 5: 'Ensino Superior completo', 9: 'Ignorado'
}

dic_necropsia = {1: 'Sim', 2: 'Não', 9: 'Ignorado', '1': 'Sim', '2': 'Não', '9': 'Ignorado'}

dic_acidtrab = {1: 'Sim', 2: 'Não', 9: 'Ignorado', '1': 'Sim', '2': 'Não', '9': 'Ignorado'}

# Aplicação dos mapeamentos básicos no DataFrame
if 'SEXO' in dt_final.columns:
    dt_final['SEXO_DESC'] = dt_final['SEXO'].map(dic_sexo).fillna('Não Informado')
if 'RACACOR' in dt_final.columns:
    dt_final['RACACOR_DESC'] = dt_final['RACACOR'].map(dic_racacor).fillna('Não Informado')
if 'LOCOCOR' in dt_final.columns:
    dt_final['LOCOCOR_DESC'] = dt_final['LOCOCOR'].map(dic_lococor).fillna('Não Informado')
if 'TIPOBITO' in dt_final.columns:
    dt_final['TIPOBITO_DESC'] = dt_final['TIPOBITO'].map(dic_tipobito).fillna('Não Informado')
if 'ESC2010' in dt_final.columns:
    dt_final['ESC2010_DESC'] = dt_final['ESC2010'].map(dic_esc2010).fillna('Não Informado')
if 'NECROPSIA' in dt_final.columns:
    dt_final['NECROPSIA_DESC'] = dt_final['NECROPSIA'].map(dic_necropsia).fillna('Não Informado')
if 'ACIDTRAB' in dt_final.columns:
    dt_final['ACIDTRAB_DESC'] = dt_final['ACIDTRAB'].map(dic_acidtrab).fillna('Não Informado')

print("Mapeamentos estáticos concluídos!")
```

### Etapa 3: Tradução da Idade (Padrão DATASUS) e Finalização
Aplicação do decodificador de idade baseado na unidade do tempo do DATASUS (horas, dias, meses e anos).

```python
def transcrever_idade_datasus(cod_idade):
    try:
        cod_str = str(int(float(cod_idade)))
        if len(cod_str) < 3:
            return "Menor de 1 ano"
        prefixo = cod_str[0]
        valor = int(cod_str[1:])
        if prefixo == '4':
            return f"{valor} Anos"
        elif prefixo == '3':
            return f"{valor} Meses"
        elif prefixo == '2':
            return f"{valor} Dias"
        elif prefixo == '1':
            return f"{valor} Horas"
        else:
            return "Idade ignorada"
    except:
        return "Não Informado"

if 'IDADE' in dt_final.columns:
    dt_final['IDADE_DESC'] = dt_final['IDADE'].apply(transcrever_idade_datasus)
else:
    dt_final['IDADE_DESC'] = 'Não Informado'

print("Processamento das idades finalizado com sucesso!")
```

```python
import os
import pandas as pd
import numpy as np

# 1. Formatar HORAOBITO para o padrão HH:MM
def formatar_hora(hora):
    try:
        if pd.isna(hora):
            return "Não Informado"
        # Converte float para string sem decimais
        h_str = str(int(float(hora))).zfill(4)
        if len(h_str) == 4:
            hh = h_str[:2]
            mm = h_str[2:]
            if 0 <= int(hh) < 24 and 0 <= int(mm) < 60:
                return f"{hh}:{mm}"
        return "Não Informado"
    except:
        return "Não Informado"

if 'HORAOBITO' in dt_final.columns:
    print("Formatando coluna HORAOBITO...")
    dt_final['HORAOBITO'] = dt_final['HORAOBITO'].apply(formatar_hora)

# 2. Traduzir CODMUNRES usando CADMUN.CNV se disponível
caminho_mun = os.path.join(pasta_tabdo, 'CADMUN.CNV')
if 'CODMUNRES' in dt_final.columns and os.path.exists(caminho_mun):
    print("Lendo dicionário de municípios CADMUN.CNV...")
    dic_mun = ler_arquivo_cnv_correto(caminho_mun)
    dt_final['CODMUNRES'] = dt_final['CODMUNRES'].astype(str).str.split('.').str[0].map(dic_mun).fillna('Não Informado')

# 3. Traduzir ACIDTRAB utilizando o dicionário ACIDTRAB.CNV
caminho_acid = os.path.join(pasta_tabdo, 'ACIDTRAB.CNV')
if 'ACIDTRAB' in dt_final.columns and os.path.exists(caminho_acid):
    print("Lendo dicionário de acidentes de trabalho ACIDTRAB.CNV...")
    dic_acid = ler_arquivo_cnv_correto(caminho_acid)
    dt_final['ACIDTRAB'] = dt_final['ACIDTRAB'].astype(str).str.split('.').str[0].map(dic_acid).fillna('Não Informado')

# 4. Traduzir as Linhas de Causas da Morte (LINHAA, LINHAB, LINHAC, LINHAD, CODMUNCOR)
# Elas contêm o código CID-10 precedido por asterisco (ex: *T792). Vamos limpar o asterisco e mapear usando o dic_cid_correto.
def traduzir_linha_cid(valor_linha, dicionario_cid):
    try:
        if pd.isna(valor_linha):
            return "Não Informado"
        # Remove o asterisco inicial se houver e pega os 3 primeiros caracteres do CID
        cid_limpo = str(valor_linha).replace('*', '').strip()[:3]
        return dicionario_cid.get(cid_limpo, "Não Informado")
    except:
        return "Não Informado"

linhas_causa = ['LINHAA', 'LINHAB', 'LINHAC', 'LINHAD', 'LINHAII']
for col_linha in linhas_causa:
    if col_linha in dt_final.columns and 'dic_cid_correto' in globals():
        print(f"Traduzindo coluna de causa: {col_linha}...")
        dt_final[col_linha] = dt_final[col_linha].apply(lambda x: traduzir_linha_cid(x, dic_cid_correto))

# 5. Visualizar resultado final consolidado
colunas_visualizar = ['DTOBITO', 'HORAOBITO', 'CODMUNRES', 'ACIDTRAB', 'LINHAA', 'LINHAB', 'LINHAC']
colunas_existentes = [c for c in colunas_visualizar if c in dt_final.columns]
display(dt_final[colunas_existentes].head(15))

# 6. Salvar base de dados atualizada de volta ao Google Drive
caminho_limpo = os.path.join(origem, 'dt_final_completo_transcrito.csv')
dt_final.to_csv(caminho_limpo, index=False, encoding='latin1')
print(f"\nSucesso! Todas as traduções finais concluídas e salvas em: {caminho_limpo}")
```
