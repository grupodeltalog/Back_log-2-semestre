```python

# ============================================================
# ETAPA 2 — PADRONIZAÇÃO DAS BASES PARA O CRUZAMENTO
# ============================================================

import pandas as pd
import numpy as np
import unicodedata
import re


# ------------------------------------------------------------
# 1. FUNÇÃO PARA NORMALIZAR TEXTOS
# ------------------------------------------------------------

def normalizar_texto(valor):

    if pd.isna(valor):
        return ''

    texto = str(valor).strip().upper()

    texto = unicodedata.normalize('NFKD', texto)
    texto = ''.join(
        caractere for caractere in texto
        if not unicodedata.combining(caractere)
    )

    texto = re.sub(r'\s+', ' ', texto)

    return texto


# ------------------------------------------------------------
# 2. PADRONIZAR MUNICÍPIOS
# ------------------------------------------------------------

prf_2['_municipio_norm'] = (
    prf_2['municipio'].map(normalizar_texto)
)

dt_final['_municipio_norm'] = (
    dt_final['CODMUNOCOR'].map(normalizar_texto)
)


# ------------------------------------------------------------
# 3. PADRONIZAR SEXO
# ------------------------------------------------------------

def normalizar_sexo(valor):

    texto = normalizar_texto(valor)

    if texto in ['MASCULINO', 'M', '1']:
        return 'M'

    if texto in ['FEMININO', 'F', '2']:
        return 'F'

    return ''


prf_2['_sexo_norm'] = prf_2['sexo'].map(normalizar_sexo)

dt_final['_sexo_norm'] = dt_final['SEXO'].map(normalizar_sexo)


# ------------------------------------------------------------
# 4. PADRONIZAR IDADE
# ------------------------------------------------------------

prf_2['_idade_num'] = pd.to_numeric(
    prf_2['idade'],
    errors='coerce'
)

dt_final['_idade_num'] = pd.to_numeric(
    dt_final['IDADE'],
    errors='coerce'
)


# ------------------------------------------------------------
# 5. PADRONIZAR DATAS
# ------------------------------------------------------------

# PRF: formato AAAA-MM-DD
prf_2['_data_acidente'] = pd.to_datetime(
    prf_2['data'],
    format='%Y-%m-%d',
    errors='coerce'
)

# DATASUS: formato DD/MM/AAAA
dt_final['_data_obito'] = pd.to_datetime(
    dt_final['DTOBITO'],
    format='%d/%m/%Y',
    errors='coerce'
)


# ------------------------------------------------------------
# 6. PADRONIZAR SITUAÇÃO FÍSICA E VEÍCULO
# ------------------------------------------------------------

prf_2['_estado_norm'] = (
    prf_2['estado_fisico'].map(normalizar_texto)
)

prf_2['_veiculo_norm'] = (
    prf_2['tipo_veiculo'].map(normalizar_texto)
)

prf_2['_envolvido_norm'] = (
    prf_2['tipo_envolvido'].map(normalizar_texto)
)

dt_final['_causabas_norm'] = (
    dt_final['CAUSABAS'].map(normalizar_texto)
)


# ------------------------------------------------------------
# 7. CONFERÊNCIA DA PADRONIZAÇÃO
# ------------------------------------------------------------

print('PADRONIZAÇÃO CONCLUÍDA')
print('\nPRF_2')
print('Total de registros:', len(prf_2))
print('Datas de acidente válidas:', prf_2['_data_acidente'].notna().sum())
print('Idades válidas:', prf_2['_idade_num'].notna().sum())
print('Municípios preenchidos:', (prf_2['_municipio_norm'] != '').sum())

print('\nDATASUS')
print('Total de registros:', len(dt_final))
print('Datas de óbito válidas:', dt_final['_data_obito'].notna().sum())
print('Idades válidas:', dt_final['_idade_num'].notna().sum())
print('Municípios preenchidos:', (dt_final['_municipio_norm'] != '').sum())


print('\nPERÍODO DOS ACIDENTES PRF:')
print(prf_2['_data_acidente'].min(), 'até', prf_2['_data_acidente'].max())

print('\nPERÍODO DOS ÓBITOS DATASUS:')
print(dt_final['_data_obito'].min(), 'até', dt_final['_data_obito'].max())


print('\nSITUAÇÕES FÍSICAS NA PRF:')
print(prf_2['estado_fisico'].value_counts(dropna=False))


print('\nIDADES DO DATASUS — CONFERÊNCIA:')
print(dt_final['_idade_num'].describe())

```

```python

# ============================================================
# ETAPA 3 — CLASSIFICAÇÃO DOS TIPOS DE TRANSPORTE
# ============================================================

# ------------------------------------------------------------
# 1. CLASSIFICAR OS VEÍCULOS DA PRF
# ------------------------------------------------------------

def classificar_prf(linha):

    envolvido = linha['_envolvido_norm']
    veiculo = linha['_veiculo_norm']

    # Pedestre: prevalece sobre o tipo de veículo
    if envolvido == 'PEDESTRE':
        return 'PEDESTRE'

    # Motocicletas e motonetas
    if veiculo in ['MOTOCICLETA', 'MOTONETA']:
        return 'MOTOCICLETA'

    # Bicicletas
    if veiculo == 'BICICLETA':
        return 'BICICLETA'

    # Ônibus e micro-ônibus
    if veiculo in ['ONIBUS', 'MICRO-ONIBUS']:
        return 'ONIBUS'

    # Veículos pesados
    if veiculo in [
        'CAMINHAO',
        'CAMINHAO-TRATOR',
        'SEMI-REBOQUE',
        'REBOQUE'
    ]:
        return 'PESADO'

    # Automóveis e veículos leves
    if veiculo in [
        'AUTOMOVEL',
        'CAMINHONETE',
        'CAMIONETA',
        'UTILITARIO'
    ]:
        return 'AUTOMOVEL'

    return 'NAO_IDENTIFICADO'


prf_2['_grupo_transporte'] = prf_2.apply(
    classificar_prf,
    axis=1
)


# ------------------------------------------------------------
# 2. CLASSIFICAR AS CAUSAS BÁSICAS DO DATASUS
# ------------------------------------------------------------

def classificar_datasus(causa):

    texto = normalizar_texto(causa)

    # Pedestres
    if 'PEDESTRE' in texto:
        return 'PEDESTRE'

    # Motociclistas
    if 'MOTOCICLISTA' in texto:
        return 'MOTOCICLETA'

    # Ciclistas
    if 'CICLISTA' in texto:
        return 'BICICLETA'

    # Ônibus
    if 'ONIBUS' in texto:
        return 'ONIBUS'

    # Veículos de transporte pesado
    if 'TRANSPORTE PESADO' in texto:
        return 'PESADO'

    # Automóveis, carros, pickups e caminhonetes
    if any(palavra in texto for palavra in [
        'AUTOMOVEL',
        'CARRO',
        'PICK UP',
        'CAMINHONETE'
    ]):
        return 'AUTOMOVEL'

    return 'NAO_IDENTIFICADO'


dt_final['_grupo_transporte'] = dt_final['CAUSABAS'].map(
    classificar_datasus
)


# ------------------------------------------------------------
# 3. CONFERIR A CLASSIFICAÇÃO DA PRF
# ------------------------------------------------------------

print('CLASSIFICAÇÃO DOS TRANSPORTES — PRF_2')

print(
    prf_2['_grupo_transporte']
    .value_counts(dropna=False)
)


# ------------------------------------------------------------
# 4. CONFERIR A CLASSIFICAÇÃO DO DATASUS
# ------------------------------------------------------------

print('\nCLASSIFICAÇÃO DOS TRANSPORTES — DATASUS')

print(
    dt_final['_grupo_transporte']
    .value_counts(dropna=False)
)


# ------------------------------------------------------------
# 5. EXAMINAR CAUSAS BÁSICAS NÃO IDENTIFICADAS
# ------------------------------------------------------------

print('\nCAUSAS BÁSICAS NÃO IDENTIFICADAS — DATASUS')

print(
    dt_final.loc[
        dt_final['_grupo_transporte'] == 'NAO_IDENTIFICADO',
        'CAUSABAS'
    ]
    .value_counts()
    .head(30)
)


# ------------------------------------------------------------
# 6. CONFERIR A DISTRIBUIÇÃO POR SITUAÇÃO FÍSICA
# ------------------------------------------------------------

print('\nSITUAÇÃO FÍSICA E GRUPO DE TRANSPORTE — PRF')

print(
    pd.crosstab(
        prf_2['estado_fisico'],
        prf_2['_grupo_transporte']
    )
)

```

```python
print('CATEGORIAS DA PRF:')
print(prf_2['_grupo_transporte'].value_counts(dropna=False))

print('\nCATEGORIAS DO DATASUS:')
print(dt_final['_grupo_transporte'].value_counts(dropna=False))


prf_busca = prf_2[
    prf_2['_estado_norm'].isin([
        'LESOES GRAVES',
        'LESOES LEVES'
    ])
].copy()

print('\nREGISTROS ELEGÍVEIS NA PRF:')
print('Total:', len(prf_busca))

print('\nPor situação física:')
print(prf_busca['_estado_norm'].value_counts())


datasus_busca = dt_final[
    (dt_final['_municipio_norm'] != '') &
    (dt_final['_sexo_norm'] != '') &
    (dt_final['_idade_num'].notna()) &
    (dt_final['_data_obito'].notna()) &
    (dt_final['_grupo_transporte'] != 'NAO_IDENTIFICADO')
].copy()

print('\nREGISTROS DATASUS COM INFORMAÇÕES VÁLIDAS:')
print('Total:', len(datasus_busca))



municipios_prf = set(prf_busca['_municipio_norm'].unique())

municipios_datasus = set(datasus_busca['_municipio_norm'].unique())

municipios_comuns = municipios_prf.intersection(municipios_datasus)

print('\nCOMPATIBILIDADE DOS MUNICÍPIOS:')
print('Municípios distintos na PRF:', len(municipios_prf))
print('Municípios distintos no DATASUS:', len(municipios_datasus))
print('Municípios encontrados nas duas bases:', len(municipios_comuns))

print('\nExemplos de municípios da PRF sem correspondência:')
print(list(municipios_prf - municipios_datasus)[:20])



categorias_prf = set(prf_busca['_grupo_transporte'].unique())

categorias_datasus = set(datasus_busca['_grupo_transporte'].unique())

print('\nCATEGORIAS PRESENTES NAS DUAS BASES:')
print(categorias_prf.intersection(categorias_datasus))

print('\nCATEGORIAS DA PRF SEM CORRESPONDÊNCIA NO DATASUS:')
print(categorias_prf - categorias_datasus)


chaves_prf = prf_busca[
    (prf_busca['_municipio_norm'] != '') &
    (prf_busca['_sexo_norm'] != '') &
    (prf_busca['_idade_num'].notna()) &
    (prf_busca['_grupo_transporte'] != 'NAO_IDENTIFICADO')
].copy()

chaves_datasus = datasus_busca.copy()

colunas_chave = [
    '_municipio_norm',
    '_sexo_norm',
    '_idade_num',
    '_grupo_transporte'
]

chaves_prf['_chave'] = list(
    zip(*(chaves_prf[c] for c in colunas_chave))
)

chaves_datasus['_chave'] = list(
    zip(*(chaves_datasus[c] for c in colunas_chave))
)

conjunto_chaves_datasus = set(chaves_datasus['_chave'].unique())

chaves_compativeis = chaves_prf['_chave'].isin(
    conjunto_chaves_datasus
)

print('\nCOMPATIBILIDADE DAS CHAVES:')
print('Registros elegíveis da PRF:', len(chaves_prf))
print('Registros com chave encontrada no DATASUS:', chaves_compativeis.sum())
print(
    'Percentual com chave encontrada:',
    round(chaves_compativeis.mean() * 100, 2),
    '%'
)
```

```python
chaves_prf['_prazo_dias'] = np.where(
    chaves_prf['_estado_norm'] == 'LESOES GRAVES',
    30,
    90
)



datasus_por_chave = {}

for chave, grupo in chaves_datasus.groupby('_chave'):

    datas = np.sort(
        grupo['_data_obito'].to_numpy(dtype='datetime64[ns]')
    )

    datasus_por_chave[chave] = datas


chaves_prf['_qtd_candidatos_temporais'] = 0

for chave, indices in chaves_prf.groupby('_chave').groups.items():

    if chave not in datasus_por_chave:
        continue

    datas_obito = datasus_por_chave[chave]

    grupo_prf = chaves_prf.loc[indices]

    datas_acidente = grupo_prf[
        '_data_acidente'
    ].to_numpy(dtype='datetime64[ns]')

    prazos = grupo_prf['_prazo_dias'].to_numpy()

    datas_limite = (
        datas_acidente +
        prazos.astype('timedelta64[D]')
    )

    # Primeiro óbito no dia do acidente ou depois
    inicio = np.searchsorted(
        datas_obito,
        datas_acidente,
        side='left'
    )

    # Último óbito dentro do prazo, incluindo o dia limite
    fim = np.searchsorted(
        datas_obito,
        datas_limite,
        side='right'
    )

    quantidades = fim - inicio

    chaves_prf.loc[
        indices,
        '_qtd_candidatos_temporais'
    ] = quantidades



total_analisado = len(chaves_prf)

com_candidato = (
    chaves_prf['_qtd_candidatos_temporais'] > 0
).sum()

sem_candidato = (
    chaves_prf['_qtd_candidatos_temporais'] == 0
).sum()

print('RESULTADO DA BUSCA TEMPORAL')

print('\nRegistros analisados:', total_analisado)

print('Com pelo menos um candidato:', com_candidato)

print('Sem candidato:', sem_candidato)

print(
    'Percentual com candidato:',
    round(com_candidato / total_analisado * 100, 2),
    '%'
)



print('\nRESULTADO POR SITUAÇÃO FÍSICA')

print(
    chaves_prf.groupby('_estado_norm')[
        '_qtd_candidatos_temporais'
    ].agg(
        registros='count',
        com_candidato=lambda x: (x > 0).sum(),
        sem_candidato=lambda x: (x == 0).sum()
    )
)



print('\nQUANTIDADE DE CANDIDATOS POR REGISTRO')

print(
    chaves_prf['_qtd_candidatos_temporais']
    .value_counts()
    .sort_index()
    .head(15)
)


multiplos = chaves_prf[
    chaves_prf['_qtd_candidatos_temporais'] > 1
]

print('\nREGISTROS COM MAIS DE UM CANDIDATO:')
print(len(multiplos))

print(
    'Percentual:',
    round(len(multiplos) / total_analisado * 100, 2),
    '%'
)
```

```python
prf_com_candidatos = chaves_prf[
    chaves_prf['_qtd_candidatos_temporais'] > 0
].copy()

print('Registros PRF com candidatos:', len(prf_com_candidatos))



datasus_grupos = {}

for chave, grupo in chaves_datasus.groupby('_chave', sort=False):

    grupo_ordenado = grupo.sort_values('_data_obito')

    datasus_grupos[chave] = {
        'datas': grupo_ordenado[
            '_data_obito'
        ].to_numpy(dtype='datetime64[ns]'),

        'indices': grupo_ordenado.index.to_numpy()
    }



pares_candidatos = []

for chave, grupo_prf in prf_com_candidatos.groupby('_chave', sort=False):

    if chave not in datasus_grupos:
        continue

    datas_obito = datasus_grupos[chave]['datas']
    indices_datasus = datasus_grupos[chave]['indices']

    for indice_prf, linha in grupo_prf.iterrows():

        data_acidente = np.datetime64(linha['_data_acidente'])

        prazo = int(linha['_prazo_dias'])

        data_limite = data_acidente + np.timedelta64(prazo, 'D')

        inicio = np.searchsorted(
            datas_obito,
            data_acidente,
            side='left'
        )

        fim = np.searchsorted(
            datas_obito,
            data_limite,
            side='right'
        )

        for posicao in range(inicio, fim):

            indice_datasus = indices_datasus[posicao]

            diferenca_dias = int(
                (datas_obito[posicao] - data_acidente)
                / np.timedelta64(1, 'D')
            )

            pares_candidatos.append((
                indice_prf,
                indice_datasus,
                diferenca_dias
            ))


df_pares = pd.DataFrame(
    pares_candidatos,
    columns=[
        'indice_prf',
        'indice_datasus',
        'dias_entre_acidente_obito'
    ]
)

print('\nPARES CANDIDATOS:', len(df_pares))



df_pares = df_pares.sort_values(
    by=[
        'dias_entre_acidente_obito',
        'indice_prf',
        'indice_datasus'
    ]
)

prf_atribuidos = set()
datasus_atribuidos = set()

atribuicoes = []

for linha in df_pares.itertuples(index=False):

    indice_prf = linha.indice_prf
    indice_datasus = linha.indice_datasus
    dias = linha.dias_entre_acidente_obito

    if indice_prf in prf_atribuidos:
        continue

    if indice_datasus in datasus_atribuidos:
        continue

    atribuicoes.append((
        indice_prf,
        indice_datasus,
        dias
    ))

    prf_atribuidos.add(indice_prf)
    datasus_atribuidos.add(indice_datasus)


df_atribuicoes = pd.DataFrame(
    atribuicoes,
    columns=[
        'indice_prf',
        'indice_datasus',
        'dias_entre_acidente_obito'
    ]
)



if not df_atribuicoes.empty:

    dados_datasus = dt_final.loc[
        df_atribuicoes['indice_datasus'].to_numpy(),
        ['ESC2010', 'OCUP']
    ].reset_index(drop=True)

    df_atribuicoes = df_atribuicoes.reset_index(drop=True)

    df_atribuicoes['ESC2010'] = dados_datasus['ESC2010']
    df_atribuicoes['OCUP'] = dados_datasus['OCUP']



print('\nRESULTADO DAS ATRIBUIÇÕES')

print('Pares candidatos:', len(df_pares))
print('Óbitos atribuídos:', len(df_atribuicoes))
print('Registros PRF contemplados:', df_atribuicoes['indice_prf'].nunique())
print('Óbitos DATASUS utilizados:', df_atribuicoes['indice_datasus'].nunique())



if not df_atribuicoes.empty:

    df_atribuicoes['_situacao_prf'] = (
        chaves_prf.loc[
            df_atribuicoes['indice_prf'],
            '_estado_norm'
        ].to_numpy()
    )

    print('\nATRIBUIÇÕES POR SITUAÇÃO FÍSICA')

    print(df_atribuicoes['_situacao_prf'].value_counts())


if not df_atribuicoes.empty:

    print('\nINTERVALO ENTRE ACIDENTE E ÓBITO')

    print(
        df_atribuicoes[
            'dias_entre_acidente_obito'
        ].describe()
    )

    print('\nFREQUÊNCIA POR INTERVALO')

    print(
        pd.cut(
            df_atribuicoes['dias_entre_acidente_obito'],
            bins=[-1, 0, 7, 15, 30, 60, 90],
            labels=[
                'Mesmo dia',
                '1 a 7 dias',
                '8 a 15 dias',
                '16 a 30 dias',
                '31 a 60 dias',
                '61 a 90 dias'
            ]
        ).value_counts().sort_index()
    )
```

```python

# ============================================================
# ETAPA FINAL — RECLASSIFICAÇÃO DE VÍTIMAS COM ÓBITO DATASUS
# ============================================================

# Criar cópia de segurança antes das alterações
prf_final_original = prf_final.copy()

# Identificar registros com correspondência potencial no DATASUS
mascara_obito = (
    prf_final['OBITO_DATASUS']
    .astype('string')
    .str.strip()
    .str.upper() == 'SIM'
)

# Converter as colunas para valores numéricos
for coluna in ['feridos_graves', 'feridos_leves', 'mortos']:
    prf_final[coluna] = pd.to_numeric(
        prf_final[coluna], errors='coerce'
    ).fillna(0)

# Identificar acidentes com exatamente um ferido
mascara_uma_vitima = (
    (prf_final['feridos_graves'] + prf_final['feridos_leves']) == 1
)

# Combinar as duas condições
mascara_reclassificacao = mascara_obito & mascara_uma_vitima

# Quantidade de registros que serão alterados
total_reclassificados = mascara_reclassificacao.sum()

# Transferir a vítima da categoria de feridos para mortos
prf_final.loc[mascara_reclassificacao, 'feridos_graves'] = 0
prf_final.loc[mascara_reclassificacao, 'feridos_leves'] = 0

prf_final.loc[mascara_reclassificacao, 'mortos'] += 1

# Relatório
print('RESULTADO DA RECLASSIFICAÇÃO')
print('Registros com OBITO_DATASUS = Sim:', mascara_obito.sum())
print('Registros com exatamente um ferido:', mascara_reclassificacao.sum())
print('Registros reclassificados:', total_reclassificados)

print('\nCONFERÊNCIA')
print(
    'Feridos graves restantes:',
    prf_final.loc[mascara_reclassificacao, 'feridos_graves'].sum()
)
print(
    'Feridos leves restantes:',
    prf_final.loc[mascara_reclassificacao, 'feridos_leves'].sum()
)
print(
    'Óbitos acrescentados:',
    prf_final.loc[mascara_reclassificacao, 'mortos'].sum()
)
```

```python

# ============================================================
# AUMENTO DE ÓBITOS POR UNIDADE FEDERATIVA
# ============================================================

# Criar cópias para a comparação
antes = prf_final_original.copy()
depois = prf_final.copy()

# Garantir que mortos seja numérico
antes['mortos'] = pd.to_numeric(
    antes['mortos'], errors='coerce'
).fillna(0)

depois['mortos'] = pd.to_numeric(
    depois['mortos'], errors='coerce'
).fillna(0)

# Somar óbitos por UF
obitos_antes = antes.groupby('uf')['mortos'].sum()
obitos_depois = depois.groupby('uf')['mortos'].sum()

# Construir tabela comparativa
comparacao_uf = pd.DataFrame({
    'Obitos_antes': obitos_antes,
    'Obitos_depois': obitos_depois
}).fillna(0)

# Calcular aumento absoluto
comparacao_uf['Aumento'] = (
    comparacao_uf['Obitos_depois'] -
    comparacao_uf['Obitos_antes']
)

# Calcular aumento percentual
comparacao_uf['Aumento_percentual'] = (
    comparacao_uf['Aumento'] /
    comparacao_uf['Obitos_antes'].replace(0, pd.NA)
) * 100

# Ordenar estados pelo maior aumento absoluto
comparacao_uf = comparacao_uf.sort_values(
    'Aumento', ascending=False
)

# Exibir resultado
print('AUMENTO DE ÓBITOS POR ESTADO')

display(comparacao_uf.round(2))

# Exibir somente estados que tiveram aumento
print('\nESTADOS COM AUMENTO DE ÓBITOS')

display(
    comparacao_uf[
        comparacao_uf['Aumento'] > 0
    ].round(2)
)
```
